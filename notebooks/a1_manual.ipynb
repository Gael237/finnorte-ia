**LSCA2318 · Tema 2.** Antes de que un modelo calcule vectores, tú los vas a construir a mano para entender qué significa «parecido = cerca».

**Qué haces:**
1. Eliges **4 a 6 dimensiones con nombre** que tengan sentido en tu variante (ej. `seguridad, pruebas, dinero, despliegue`).
2. Escribes **3 frases** de tu variante: F1 y F2 del mismo tema, F3 de otro tema.
3. Les das un valor de 0 a 1 a cada dimensión (justifícalo en una línea).
4. Normalizas y calculas el coseno F1–F2 y F1–F3. **Interpretas** el resultado.

Funciona en Colab o en VS Code. No necesita internet ni llave.

```python
import numpy as np

# 1) Tus dimensiones (4 a 6). Cámbialas por las de tu variante.
dimensiones = ["memoria", "red", "seguridad", "dinero"]

# 2) y 3) Tus frases y sus vectores (un número de 0 a 1 por dimensión, en el mismo orden).
frases = {
    "F1": ("La aplicación se cerró repentinamente por falta de RAM",                              [0.9, 0.1, 0.0, 0.0]),
    "F2": ("Tenemos un error crítico de OOM (Out of Memory) en el servidor",                      [0.9, 0.2, 0.0, 0.0]),
    "F3": ("El certificado TLS expiró y los usuarios no tienen conexión segura",                  [0.0, 0.8, 0.9, 0.0]),
}
for k, (texto, v) in frases.items():
    assert len(v) == len(dimensiones), f"{k} debe tener {len(dimensiones)} valores"
    print(k, texto, dict(zip(dimensiones, v)))
```

**Justificación de tus valores** (doble clic para editar): explica en una línea por qué F1 tiene 0.9 en «seguridad», etc.

- F1: Mucha memoria, casi nada de red, cero seguridad.
- F2: Mucha memoria (OOM), un poco de red (servidor).
- F3: Cero de memoria, mucha red (conexión), mucha seguridad (TLS).

```python
def normalizar(v):
    v = np.asarray(v, dtype=float)
    n = np.linalg.norm(v)
    return v / n if n > 0 else v      # el vector cero se queda en cero

def coseno(a, b):
    a, b = normalizar(a), normalizar(b)
    return float(a @ b)

V = {k: v for k, (_, v) in frases.items()}
for k in V:
    print(f"{k} normalizado:", np.round(normalizar(V[k]), 3), " norma =", round(np.linalg.norm(normalizar(V[k])), 3))

print(f"\ncos(F1, F2) = {coseno(V['F1'], V['F2']):.3f}   ← mismo tema")
print(f"cos(F1, F3) = {coseno(V['F1'], V['F3']):.3f}   ← temas distintos")
```

**Interpretación** (doble clic para editar, 4–6 líneas):
1. ¿Salió F1–F2 mayor que F1–F3? ¿Por qué?
2. ¿Qué dimensión pesó más en el resultado?
3. Un modelo real usa cientos de dimensiones **sin nombre**. ¿Qué ganas y qué pierdes frente a tus dimensiones con nombre?

Al calcular el coseno, observamos que la similitud entre la Frase 1 y la Frase 2 es muy cercana a 1. Esto se debe a que ambas comparten
una alta carga semántica en la dimensión de "memoria" (RAM y OOM significan lo mismo en este contexto). 
Por el contrario, la similitud entre la Frase 1 y la Frase 3 es cercana a 0, ya que sus vectores son ortogonales; una trata exclusivamente
sobre infraestructura de memoria, mientras que la otra trata sobre seguridad de red (certificados TLS).
